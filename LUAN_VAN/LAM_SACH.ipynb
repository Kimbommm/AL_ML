```python
import os
import pandas as pd
```

```python


# Đường dẫn tới thư mục cha chứa các thư mục con và file Excel
root_folder = "C:/Users/admin/AL_ML/DU_LIEU"

# Tạo một danh sách để lưu trữ các DataFrame từ các file Excel
dataframes = []

# Duyệt qua tất cả các thư mục con và file trong thư mục cha
for folder_name, subfolders, filenames in os.walk(root_folder):
    for filename in filenames:
        if filename.endswith('.xlsx') or filename.endswith('.xls'):  # Kiểm tra file có phải là file Excel không
            file_path = os.path.join(folder_name, filename)  # Tạo đường dẫn đầy đủ tới file
            print(f"Đang đọc file: {file_path}")
            
            # Đọc file Excel và lưu vào danh sách
            df = pd.read_excel(file_path)
            dataframes.append(df)

# Nếu muốn gộp tất cả các file Excel lại thành một DataFrame duy nhất
combined_df = pd.concat(dataframes, ignore_index=True)

# Kiểm tra dữ liệu đã đọc
print(combined_df.head())
```

```python
# Xuất DataFrame ra file Excel
output_file_path = "C:/Users/admin/AL_ML/DU_LIEU/combined_output.xlsx"
combined_df.to_excel(output_file_path, index=False)  # index=False để không lưu cột chỉ số vào file Excel

print(f"File đã được lưu thành công tại: {output_file_path}")
```

Đoạn mã Python để xử lý nhiều file và lưu dưới dạng Excel

```python
import pandas as pd
import glob
import os
```

```python
# Đường dẫn đến thư mục chứa các file CSV
input_directory = r'C:\Users\admin\AL_ML\DU_LIEU\대전시 버스 교통카드거래 이력 데이터\*.csv'
# Đường dẫn đến thư mục để lưu các file Excel đã chỉnh sửa
output_directory = r'C:\Users\admin\AL_ML\DU_LIEU\대전시 버스 교통카드거래 이력 데이터\output'

# Tạo thư mục đầu ra nếu nó chưa tồn tại
os.makedirs(output_directory, exist_ok=True)

# Lấy tất cả các file CSV trong thư mục
file_list = glob.glob(input_directory)

for file in file_list:
    # Đọc file CSV với dấu phân cách là '|'
    df = pd.read_csv(file, sep='|')
    
    # Tạo tên file đầu ra với định dạng .xlsx
    output_file = os.path.join(output_directory, os.path.basename(file).replace('.csv', '.xlsx'))  # Thay đổi đuôi file sang .xlsx
    
    # Xuất lại DataFrame thành file Excel
    df.to_excel(output_file, index=False, engine='openpyxl')  # Hoặc bạn có thể sử dụng engine='xlsxwriter'

print("Đã hoàn thành việc chuyển đổi các file trong thư mục và lưu dưới dạng Excel!")
```

Đoạn mã giải nén nhiều file cùng 1 lúc

```python
import zipfile
import glob
import os
```

```python


# Đường dẫn đến thư mục chứa các file ZIP
input_directory = r'C:\Users\admin\AL_ML\DU_LIEU\대전시 대중교통 이용정보 데이터\Lan_1\*.zip'
# Thư mục để lưu các file đã giải nén
extract_to = r'C:\Users\admin\AL_ML\DU_LIEU\대전시 대중교통 이용정보 데이터\output'

# Tạo thư mục đầu ra nếu nó chưa tồn tại
os.makedirs(extract_to, exist_ok=True)

# Lấy tất cả các file ZIP trong thư mục
zip_files = glob.glob(input_directory)

# Biến để theo dõi số lượng file đã giải nén
file_count = 1

if not zip_files:
    print("Không tìm thấy file ZIP nào trong thư mục đã chỉ định.")
else:
    for zip_file in zip_files:
        print(f"Đang giải nén file: {zip_file}")
        # Giải nén từng file ZIP
        with zipfile.ZipFile(zip_file, 'r') as zip_ref:
            # Lấy danh sách tất cả các file trong ZIP
            for member in zip_ref.namelist():
                # Đặt tên mới cho file dựa trên số lượng đã giải nén
                new_file_name = f"{file_count}{os.path.splitext(member)[-1]}"  # Giữ nguyên đuôi file gốc
                output_file_path = os.path.join(extract_to, new_file_name)
                
                # Giải nén file và lưu với tên mới
                with open(output_file_path, 'wb') as output_file:
                    output_file.write(zip_ref.read(member))
                print(f"Đã giải nén file: {new_file_name}")
                file_count += 1  # Tăng biến đếm số lượng file

    print("Đã hoàn thành việc giải nén và đặt tên cho tất cả các file!")
```
